### Looping in a list

```python
a= int(input("Enter your score: "))
```

```python
if a>=50:
  print("Pass")
else:
  print("fail")
```

```python
Scores= [50,70, 100, 120, 80]
```

```python
if Scores >= 100:
  print("A")
```

```python
for i in Scores:
  if i >= 100:
    print(f"{i} is A")
  elif i >= 80:
    print(f"{i} is B")
  elif i >= 60:
    print(f"{i} is C")
  else:
    print(f"{i} is F")
```
